

### **Python Logging**

Logging is a crucial aspect of any application, providing a way to track events, errors, and operational information. Python's built-in logging module offers a flexible framework for emitting log messages from Python programs. In this lesson, we will cover the basics of logging, including how to configure logging, log levels, and best practices for using logging in Python applications.

```python
import logging
```

```python
# building basic logging level
logging.basicConfig(level=logging.DEBUG)
```

```python
# different severity message
logging.debug("This is the debug message")
logging.warning("This is the warning message")
logging.error("This is the error message")
logging.info("This is the info message")
logging.critical("This is the critical message")
```



### **Log Levels**

Python's logging module has several log levels indicating the severity of events. The default levels are:

* DEBUG: Detailed information, typically of interest only when diagnosing problems.
* INFO: Confirmation that things are working as expected.
* WARNING: An indication that something unexpected happened or indicative of some problem in the near future (e.g., 'disk space low'). The software is still working as expected.
* ERROR: Due to a more serious problem, the software has not been able to perform... *(cut off)*
* CRITICAL: A very serious error, indicating that the program itself may be unable... *(cut off)*

```python
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
logging.debug("This is the debug message")
logging.warning("This is the warning message")
logging.error("This is the error message")
logging.info("This is the info message")
logging.critical("This is the critical message")
```

## logging with multiple Modules

```python
import logging

# 1. Clear any existing notebook log configuration
for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)

logger1=logging.getLogger("Module1") # giving the name to the logger
logger1.setLevel(logging.DEBUG) # setting the level

logger2=logging.getLogger("Module2") # same as above
logger2.setLevel(logging.WARNING)

# Basic logging configure
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
```

```python
logger1.debug("This is the debug message")
logger2.warning("This is the critical Message")
logger2.critical("This is the critical Message")
```

## A real World Example

```python
import logging

for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)
    
logging.basicConfig(
     level=logging.DEBUG,
     format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
     datefmt='%Y-%m-%d %H:%M:%S',
     # when using handler no need to write filename and filemode ok
     handlers=[ 
        logging.FileHandler('app1.log',mode='w'),
        logging.StreamHandler()
    ]
)

logger=logging.getLogger("Arithimatic_App")

def add(a,b):
    result=a + b
    logger.debug(f"Addition of {a} + {b}= {result}")
    return result

def subtract(a,b):
    result=a - b
    logger.debug(f"Subtraction of {a} - {b}= {result}")
    return result

def multiply(a,b):
    result=a * b
    logger.debug(f"Multiplication of {a} * {b}= {result}")
    return result

def division(a,b):
    try:
        result=a / b
        logger.debug(f"Addition of {a} + {b}= {result}")
        return result
    except ZeroDivisionError :
        logger.debug("Can't divide by 0")
        return None
# now calling the function
add(10,25)
subtract(34,16)
multiply(4,8)
division(56,8)
```
