```python
to_do_list = []
continue_program = True

while continue_program:
    print("\n--- TO-DO MENU ---")
    print("1: Add | 2: Complete | 3: View | 4: Exit")
    
    try:
        user_choice = int(input("Enter your choice: "))
    except ValueError:
        print("Error: Please enter a number (1-4) only.")
        continue # Skips the rest of the loop and starts over

    if user_choice == 1:
        user_input = input("Enter the task: ").strip()
        if user_input:
            to_do_list.append(user_input)
        else:
            print("Task cannot be empty!")
        
    elif user_choice == 2:
        done_task = input("Enter the task you completed: ").strip()
        if done_task in to_do_list:
            to_do_list.remove(done_task)
            print(f"Removed: {done_task}")
        else:
            print("Task not found in list.")
            
    elif user_choice == 3:
        if not to_do_list:
            print("Your list is currently empty.")
        else:
            print("\nPending Tasks:")
            for index, task in enumerate(to_do_list, start=1):
                print(f"{index}. {task}")
                
    elif user_choice == 4:
        print("Goodbye!")
        continue_program = False
    else:
        print("Invalid choice. Choose 1-4.")
```
