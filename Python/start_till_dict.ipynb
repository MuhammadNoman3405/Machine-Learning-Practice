#### >>>>>>>>> Merge the Dict>>>>>>>>>>>>

```python
# Dictionary 1: Personal Details
personal_info = {
    "name": "Noman",
    "major": "CS",
    "city": "Multan"
}

# Dictionary 2: Academic Achievements
academic_info = {
    "cgpa": 3.66,
    "semester": 6,
    "skills": ["Python", "Machine Learning"]
}

merged_dict={**personal_info,**academic_info} 
 
for keys,values in merged_dict.items():
    print(f"{keys} : {values}")
```

#### Way 2nd by | operator

```python
# Dictionary 1: Personal Details
personal_info = {
    "name": "Noman",
    "major": "CS",
    "city": "Multan"
}

# Dictionary 2: Academic Achievements
academic_info = {
    "cgpa": 3.66,
    "semester": 6,
    "skills": ["Python", "Machine Learning"]
}

merged_dict=personal_info | academic_info  ## also used by using merge operator | 
 
for keys,values in merged_dict.items():
    print(f"{keys} : {values}")
```

#### 3rd way by using the .update()

```python
# Dictionary 1: Personal Details
personal_info = {
    "name": "Noman",
    "major": "CS",
    "city": "Multan"
}

# Dictionary 2: Academic Achievements
academic_info = {
    "cgpa": 3.66,
    "semester": 6,
    "skills": ["Python", "Machine Learning"]
}

personal_info.update(academic_info)
 
for keys,values in personal_info.items():
    print(f"{keys} : {values}")
```
