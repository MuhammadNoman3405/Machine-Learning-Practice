## Json Data Serilization

```python
# json taks
import json
data={'name':'Muhammad Noman'
      ,'Age':19,
      'Address':'Multan'}

# dict to json string

json_str=json.dumps(data)
print(json_str)
print(type(json_str))

# now json to reverse dict conversion

parser=json.loads(json_str)
print(parser)
print(type(parser))
```

```python
import csv

with open('example.csv',mode='w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age']) # row1 data
    writer.writerow(['Muhammad Noman',19])
    
with open('example.csv',mode='r') as file:
    Reader=csv.reader(file)
    for i in Reader:
        print(i)
```

#### Datetime

```python
## datetime
from datetime import datetime,timedelta
now=datetime.now()
print(now)

yesterday=now - timedelta(days=1)
print(yesterday)
```

```python
# Regular expression
import re

pattern=r'\d+'
text='i won this game 10 times'

match=re.search(pattern,text) # this search the matched tings in the text ok 
print(match.group())
```

```python
with open('readfile.txt','r')  as file:
    content=file.read() # this fetches the whole contents.
    print(content)
```

```python
with open('readfile.txt','r') as file:
    for line in file:
        print(line)
```

```python
# to remove the newline write .stripe()
with open('readfile.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
# write with the append 
with open('readfile.txt','a') as file:
    file.write('\n Am appending the file')
```

```python
# with the list add to the files

list=['First line\n','Second line\n','Third line\n']
with open('readfile.txt','a') as file:
    file.writelines(list)
```
