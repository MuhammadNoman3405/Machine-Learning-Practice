#### leap year

```python
## leap year
year=int(input("Enter a year:"))
if year%4==0:
    if year%100:
        if year % 400 ==0:
            print("year is a leap year")
        else:
            print("year is not a leap year")
    else:
        print("year is a leap year")
else:
    print("year is not a leap year")
```

```python

## calculator
num1=float(input("Enter the first number:"))
num2=float(input("Enter the second number:"))
operations=input("Enter the operations( + , -  , / ,  * ):")

if operations=="+":
    result=num1+num2
elif operations=="-":
    result=num1-num2
elif operations=="/":
    if num2 ==0:
        print("Can not divide by the zero")
    else:
        result=num1/num2
elif operations=="*":
    result=num1*num2
else:
    print("the invalid input oprtations")

print(f"the result of operation {operations} is : {result}")
```

#### While Loop

```python
count=True
list=[]
while count:
    
    num=int(input("Enter the Number"))
    if num==0:
        count=False
        break
    else:
        list.append(num)
print(f"The Entered Numbers Are:{list}")
```

#### Sum of first 10 Natural Numbers viva While LOOP

```python
num=10
sum=0
count=1
while count<=num:
    sum=sum + count
    count=count + 1
    
print("The sum is :",sum)
    
    
```
