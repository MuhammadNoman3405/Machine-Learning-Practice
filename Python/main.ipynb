#### leap year

```python
## leap year
year=int(input("Enter a year:"))
if year%4==0:
    if year%100:
        if year % 400 ==0:
            print("year is a leap year")
        else:
            print("year is not a leap year")
    else:
        print("year is a leap year")
else:
    print("year is not a leap year")
```

```python

## calculator
num1=float(input("Enter the first number:"))
num2=float(input("Enter the second number:"))
operations=input("Enter the operations( + , -  , / ,  * ):")

if operations=="+":
    result=num1+num2
elif operations=="-":
    result=num1-num2
elif operations=="/":
    if num2 ==0:
        print("Can not divide by the zero")
    else:
        result=num1/num2
elif operations=="*":
    result=num1*num2
else:
    print("the invalid input oprtations")

print(f"the result of operation {operations} is : {result}")
```

#### While Loop

```python
count=True
list=[]
while count:
    
    num=int(input("Enter the Number"))
    if num==0:
        count=False
        break
    else:
        list.append(num)
print(f"The Entered Numbers Are:{list}")
```

#### List Comprehension
Basic Syntax   [expresssion for i in iterable]
with conditional logic [expresssion for i in iterable if condition]
Nested List Logic [expresssion for i in iterable1 for j in iterable2 ]

```python
## similar cab be done viva this
[x**2 for x in range(10)]
```

```python
list=[]
for i in range (10):
    if i%2==0:
        list.append(i)
print(list)
```

```python
### similar can be performed viva list comprehension
var=[num for num in range (10) if num%2==0]
print(var)
```

```python
## now nested list comprehension
list1=[1,2,3,4]
list2=['a','b','d','e']
var=[[i,j] for i in list1 for j in list2]
print(var)
```

```python
list1=[]
list2=['Apple','Banana','Grapes','Orange']
for i in list2:
    list1.append(len(i))
print(list1)
```

```python
## similar can be done viva
## list comprhension by function call

var=[len(word) for word in list2]
print(var)
```
