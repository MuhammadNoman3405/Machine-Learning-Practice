```python
# code to check the path is file or the directory?
path1='count.txt'
if os.path.isfile(path1):
    print(f"path {path1} is a file")
elif os.path.isdir(path1):
    print(f"path {path1} is a directory")
else:
    print(f"path {path1} is neither a file nor a directory")
```
