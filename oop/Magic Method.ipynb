

### Magic Methods

Magic methods in Python, also known as dunder methods (double underscore methods), are special methods that start and end with double underscores. These methods enable you to define the behavior of objects for built-in operations, such as arithmetic operations, comparisons, and more.

### Magic Methods

Magic methods are predefined methods in Python that you can override to change the behavior of your objects. Some common magic methods include:

* **init**: Initializes a new instance of a class.
* **str**: Returns a string representation of an object.
* **repr**: Returns an official string representation of an object.

```python
# let say check the magic methods of a class Person
class Person:
    pass
prsn=Person()
dir(prsn) # now see the magic methods of an object
```

```python
# without calling the dunder method
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
person=Person("Muhammad Noman",22)
print(person) # this will show mem location of the object
```

```python

#  calling the dunder method __str__
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
    def __str__(self): # dunder or magic method
        return f"Name:{self.name}\nAge:{self.age}"  
      
person=Person("Muhammad Noman",22)
print(person) # this will show string output
```

```python

#  calling the dunder method __str__
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
    # def __str__(self): # dunder or magic method
    #     return f"Name:{self.name}\nAge:{self.age}"  
    
    def __repr__(self): # official representation of the string
        return f"person(Name: {self.name},Age: {self.age})"
      
person=Person("Muhammad Noman",22)
print(person) # this will show string output
```

```python
print(repr(person)) # can also be called viva this
```

```python
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y

    # Defines the behavior of the '+' operator
    def __add__(self,param):
        if isinstance(param,Vector):
            return Vector(self.x + param.x, self.y + param.y)
        else:
            return NotImplemented
            

    # Beautiful representation for debugging
    def __repr__(self):
        return f"Vecort({self.x},{self.y})"

# Practice
v1=Vector(5,8)
v2=Vector(10,15)
v3 = v1 + v2  # This internally calls v1.__add__(v2)
print(v3)     
```

## 2. Comparisons (__eq__, __lt__)

```python
class Book:
    def __init__(self,name,pages):
        self.name=name
        self.pages=pages
    
    # Defines '==' behavior
    
    def __eq__(self,others):
        if isinstance(others,Book):
            return self.pages==others.pages # this will give us the true value
        else:
            return False
        
    # Defines '<' behavior (Python automatically uses this to infer '>' as well!)
    def __lt__(self,others):
        if isinstance(others,Book):
            return self.pages < others.pages # this will return True
        else:
            return False

# Practice
book1=Book("Data Science pro",500)
book2=Book("Ai for everyone",600)
print(book1 == book2)  # False (Calls book1.__eq__(book2))
print(book1 < book2)   # True  (Calls book1.__lt__(book2))
print(book1 > book2)   # False  (Python flips the logic of __lt__)
```

```python
class Team:
    def __init__(self,name,members):
        self.name=name
        self.members=members
        
    def __len__(self):
        return len(self.members)
    
    def __getitem__(self, position):
        return self.members[position]
    
obj=Team("Ai developers",["Muhammad","junaid","Rehan","Muneeb"])
print(len(obj))
print(obj[2])
```
