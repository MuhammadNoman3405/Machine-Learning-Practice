```python
class BankAccount():
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        
    def DepositAmount(self, amount):
        self.balance += amount
        return self.balance
        
    def WithDraw(self, amount):
        if amount > self.balance:
            return False  # Failed transaction
        self.balance -= amount
        return self.balance

# --- Handling inputs and logic inside a master function ---
def run_atm():
    name = input("Enter your name: ")
    initial_deposit = float(input("Enter initial deposit amount: "))
    
    # Initialize account
    user_account = BankAccount(name, initial_deposit)
    print(f"Account created for {user_account.owner}. Balance: {user_account.balance}")
    
    # Process withdrawal
    withdraw_amount = float(input("Enter amount to withdraw: "))
    updated_balance = user_account.WithDraw(withdraw_amount)
    
    if updated_balance is False:
        print("Transaction Declined: Insufficient Funds.")
    else:
        print(f"Transaction Successful! Updated Balance: {updated_balance}")

# Run the system
run_atm()
```

```python
class BankAccount():
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        
    def DepositAmount(self, amount):
        self.balance += amount
        print(f" Successfully deposited: {amount}. Total balance: {self.balance}")
        
    def WithDraw(self, amount):
        if amount > self.balance:
            print("Balance is insufficient. Can't withdraw.")
        else:
            self.balance -= amount
            print(f" Withdrew: {amount}. Remaining Balance: {self.balance}")
            
    def get_balance(self):
        print(f"Total Balance is: {self.balance}")

# --- User Interaction Layout ---
user_name = input("Enter your name to open an account:")
# Create the account with a starting balance of 0
account = BankAccount(user_name, balance=0)

while True:
    print("\n--- Bank Menu ---")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")
    
    choice = input("Choose an option (1-4): ")
    
    if choice == '1':
        account.get_balance()
    elif choice == '2':
        amount = float(input("Enter amount to deposit: "))
        account.DepositAmount(amount)
    elif choice == '3':
        amount = float(input("Enter amount to withdraw: "))
        account.WithDraw(amount)
    elif choice == '4':
        print(f"Thank you for banking with us, {account.owner}! Goodbye.")
        break
    else:
        print("Invalid choice. Please select a valid option.")
```
