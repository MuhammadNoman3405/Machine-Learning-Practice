### ATM System Using a Function

Wrapping account creation and a withdrawal flow inside a single `run_atm()` function, showing how objects and functions work together.

```python
class BankAccount():
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        
    def DepositAmount(self, amount):
        self.balance += amount
        return self.balance
        
    def WithDraw(self, amount):
        if amount > self.balance:
            return False  # Failed transaction
        self.balance -= amount
        return self.balance

# --- Handling inputs and logic inside a master function ---
def run_atm():
    name = input("Enter your name: ")
    initial_deposit = float(input("Enter initial deposit amount: "))
    
    # Initialize account
    user_account = BankAccount(name, initial_deposit)
    print(f"Account created for {user_account.owner}. Balance: {user_account.balance}")
    
    # Process withdrawal
    withdraw_amount = float(input("Enter amount to withdraw: "))
    updated_balance = user_account.WithDraw(withdraw_amount)
    
    if updated_balance is False:
        print("Transaction Declined: Insufficient Funds.")
    else:
        print(f"Transaction Successful! Updated Balance: {updated_balance}")

# Run the system
run_atm()
```

### Interactive Bank Menu System

A full menu-driven ATM simulation combining everything above: deposits, withdrawals, balance checks, and a clean exit — all wrapped in a `while True` loop.

```python
class BankAccount():
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        
    def DepositAmount(self, amount):
        self.balance += amount
        print(f" Successfully deposited: {amount}. Total balance: {self.balance}")
        
    def WithDraw(self, amount):
        if amount > self.balance:
            print("Balance is insufficient. Can't withdraw.")
        else:
            self.balance -= amount
            print(f" Withdrew: {amount}. Remaining Balance: {self.balance}")
            
    def get_balance(self):
        print(f"Total Balance is: {self.balance}")

# --- User Interaction Layout ---
user_name = input("Enter your name to open an account:")
# Create the account with a starting balance of 0
account = BankAccount(user_name, balance=0)

while True:
    print("\n--- Bank Menu ---")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")
    
    choice = input("Choose an option (1-4): ")
    
    if choice == '1':
        account.get_balance()
    elif choice == '2':
        amount = float(input("Enter amount to deposit: "))
        account.DepositAmount(amount)
    elif choice == '3':
        amount = float(input("Enter amount to withdraw: "))
        account.WithDraw(amount)
    elif choice == '4':
        print(f"Thank you for banking with us, {account.owner}! Goodbye.")
        break
    else:
        print("Invalid choice. Please select a valid option.")
```

## Inheritance In Python

Inheritance is a fundamental concept in Object-Oriented Programming (OOP) that allows a class to inherit attributes and methods from another class. This lesson covers single inheritance and multiple inheritance, demonstrating how to create and use them in Python.

## Polymorphism
Polymorphism is a core concept in Object-Oriented Programming (OOP) that allows objects of different classes to be treated as objects of a common superclass. It provides a way to perform a single action in different forms. Polymorphism is typically achieved through method overriding and interfaces

<a id="method-overriding"></a>
## Method Overriding
Imagine a parent class called Animal that has a method named makeSound(). By default, the animal just makes a generic noise.

If you create a child class called Dog, it inherits makeSound(). However, a generic noise doesn't fit a dog. So, the Dog class overrides the makeSound() method to print "Woof!". Similarly, a Cat class would override it to print "Meow!".

```python
# let take an example of the method overriding
#base class 
class Animal:
    def __init__(self,name):
        self.name=name
        
    def make_sound(self):
        return "A specific sound of"
    
# Derived class that will override the make_sound function
class Dog(Animal):
    
    def __init__(self,name):
        super().__init__(name)
        
    def make_sound(self):
        return f"{super().make_sound()} {self.name} is: woof"
     
# Derived class that will override the make_sound function   
class Cat(Animal):
    def __init__(self,name):
        super().__init__(name)
        
    def make_sound(self):
        return f"{super().make_sound()} {self.name} is: Meow"
    
# making a function that demonstrate ploymorphisim

def animal_speak(animal):
    return animal.make_sound()
  
# making objects  
dg=Dog('lucy')
ct=Cat('Mano')

print(dg.make_sound())
print(f"{ct.make_sound()}\n")
print("Now Ploymorphisim function output:\n")
print(animal_speak(dg))
print(animal_speak(ct))
```

```python
# let take another example
class Shape:
    def  __init__(self,shape_name):
        self.shape_name=shape_name
        
    def area(self):
        return f"The area of the"
    
class Rectangle(Shape):
    def __init__(self,shape_name,length,width):
        super().__init__(shape_name)
        self.length=length
        self.width=width
        
    def area(self):
        return f"{super().area()} {self.shape_name} is:{self.length * self.width}"
    
class Square(Shape):
    def __init__(self,shape_name,radius):
        super().__init__(shape_name)
        self.radius=radius
        
    def area(self):
        return f"{super().area()} {self.shape_name} is:{3.1415 * self.radius**2}"
    
# Now polyporphisim
def shape_area(parameter):
    print(f"{parameter.area()}")

# now let make objects
rec=Rectangle("Rectangle",4,5)
sqr=Square("Square",6)

# now call th polymorphisim finction
shape_area(rec)
shape_area(sqr)
```

## Polymorphism with Abstract Base Classes

Abstract Base Classes (ABCs) are used to define common methods for a group of related objects. They can enforce that derived classes implement particular methods, promoting consistency across different implementations.

```python
# Now let take another example by using the ABC
from abc import ABC,abstractmethod
# Base Class
class Vehical(ABC):
    @abstractmethod
    def __init__(self,vehical_name):
        self.vehical_name=vehical_name
        
    def engine_start(self):
        return f"Engine has been started"
    
# derived class
class Car(Vehical):
    def __init__(self,vehical_name):
        super().__init__(vehical_name)
        
    def engine_start(self):
        return f"{self.vehical_name} {super().engine_start()}"

class Motorcycle(Vehical):
    def __init__(self,vehical_name):
        super().__init__(vehical_name)
        
    def engine_start(self):
        return f"{self.vehical_name} {super().engine_start()}"
    
# Now Polymorphisim function
def engine_information(param):
    print(f"{param.engine_start()}")
    
# now let make the objet
cr=Car("BMW")
mtr=Motorcycle("Honda Prider")
# now call the function of the polymorphisim
engine_information(cr)
engine_information(mtr)
```

## Encapsulation And Abstraction

Encapsulation and abstraction are two fundamental principles of Object-Oriented Programming (OOP) that help in designing robust, maintainable, and reusable code. Encapsulation involves bundling data and methods that operate on the data within a single unit, while abstraction involves hiding complex implementation details and exposing only the necessary features.

### Encapsulation

Encapsulation is the concept of wrapping data (variables) and methods (functions) together as a single unit. It restricts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data.

**Abstraction**

Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object. This helps in reducing programming complexity and effort.

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    def drive(self):
        print("This is the drive function of the base class")
        
    @abstractmethod
    def engine_start(self): # just define 
        pass
    
class Car(Vehicle):
    def engine_start(self): # expand here and that is abstraction
        print("The engine has been started")
        
def car_info(param):
    param.engine_start()
    param.drive()

obj=Car()
car_info(obj) 
```

```python
from abc import ABC ,abstractmethod

class PaymentProcessor(ABC):
    def __init__(self,name):
        self._name=name
        
    def receipt(self,amount):
        print(f"Receipt generated for the amount ${amount}")
        
    @abstractmethod
    def process_payment(self,amount):
        pass
    
# now here abstract fuction will be open 
 
class CreditCardPayment(PaymentProcessor):
    def __init__(self,name):
        super().__init__(name)
        
    def process_payment(self, amount):
        print(f"{self._name} Securely processing ${amount} via Credit Card Gateway...")
        
class PayPalPayment(PaymentProcessor):
    def __init__(self,name):
        super().__init__(name)
        
    def process_payment(self, amount):
        print(f"{self._name} Securely processing ${amount} via Pay Pal Gateway...")
        
# polymorphisim function applying
def info(param,amount):
    param.process_payment(amount)
    param.receipt(amount)
    

card_obj=CreditCardPayment("Muhammad")
paypal=PayPalPayment("Muhammad")
info(card_obj,2500)
print("-" *20)
info(paypal,200)
```

### Magic Methods

Magic methods in Python, also known as dunder methods (double underscore methods), are special methods that start and end with double underscores. These methods enable you to define the behavior of objects for built-in operations, such as arithmetic operations, comparisons, and more.

### Magic Methods

Magic methods are predefined methods in Python that you can override to change the behavior of your objects. Some common magic methods include:

* **init**: Initializes a new instance of a class.
* **str**: Returns a string representation of an object.
* **repr**: Returns an official string representation of an object.

## Operator Overloading with `__add__`

Defining `__add__` lets the `+` operator work directly on custom objects, like adding two vectors.

```python
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y

    # Defines the behavior of the '+' operator
    def __add__(self,param):
        if isinstance(param,Vector):
            return Vector(self.x + param.x, self.y + param.y)
        else:
            return NotImplemented
            

    # Beautiful representation for debugging
    def __repr__(self):
        return f"Vecort({self.x},{self.y})"

# Practice
v1=Vector(5,8)
v2=Vector(10,15)
v3 = v1 + v2  # This internally calls v1.__add__(v2)
print(v3)     
```

## 2. Comparisons (__eq__, __lt__)

```python
class Book:
    def __init__(self,name,pages):
        self.name=name
        self.pages=pages
    
    # Defines '==' behavior
    
    def __eq__(self,others):
        if isinstance(others,Book):
            return self.pages==others.pages # this will give us the true value
        else:
            return False
        
    # Defines '<' behavior (Python automatically uses this to infer '>' as well!)
    def __lt__(self,others):
        if isinstance(others,Book):
            return self.pages < others.pages # this will return True
        else:
            return False

# Practice
book1=Book("Data Science pro",500)
book2=Book("Ai for everyone",600)
print(book1 == book2)  # False (Calls book1.__eq__(book2))
print(book1 < book2)   # True  (Calls book1.__lt__(book2))
print(book1 > book2)   # False  (Python flips the logic of __lt__)
```

## Container Emulation (`__len__`, `__getitem__`)

Making a custom object behave like a list — supporting `len()`, indexing, and even slicing.

```python
class Team:
    def __init__(self,name,members):
        self.name=name
        self.members=members
        
    def __len__(self):
        return len(self.members)
    
    def __getitem__(self, position):
        return self.members[position]
    
obj=Team("Ai developers",["Muhammad","junaid","Rehan","Muneeb"])
print(len(obj))
print(obj[2])
```

## Building a Vector Class with Full Operator Overloading

A complete `Vector` class implementing `+`, `-`, `*`, `/`, `==`, `<`, `>`, and `repr()` — the core toolkit of operator overloading in one example.

```python
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
        
    def __add__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x + other.x, self.y + other.y)
        else:
            return NotImplemented
            
    def __sub__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x - other.x, self.y - other.y)
        else:
            return NotImplemented
            
    def __mul__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x * other.x, self.y * other.y)
        else:
            return NotImplemented
            
    def __truediv__(self,other):
        if isinstance(other,Vector):
            if other.x!=0 and other.y !=0:
                x_part=self.x / other.x
                y_part=self.y / other.y
                return Vector(x_part, y_part)
            else:
                print("can't divide as one of the vectors y has 0 value")
        else:
            return NotImplemented
               
    def __eq__(self,other):
        if isinstance(other,Vector):
            return self.x == other.x and self.y == other.y
        else:
            return False
            
    def __lt__(self,other):
        if isinstance(other,Vector):
            return self.x < other.x and self.y < other.y
        else:
            return False
            
    def __gt__(self,other):
        if isinstance(other,Vector):
            return self.x > other.x and self.y > other.y
        else:
            return False
            
    def __repr__(self):
        return f"Vector({self.x},{self.y})"
    
 # making objects and performing tasks   
v5=Vector(2,4)
v6=Vector(6,8)
print(f"Adding Two Vectors:\n{repr(v5)} + {repr(v6)}: {v5 + v6}\n{"-"*20}")
print(f"Substracting Two Vectors:\n{repr(v5)} - {repr(v6)}: {v5 - v6}\n{"-"*20}")
print(f"Multiplying Two Vectors:\n{repr(v5)} * {repr(v6)}: {v5 * v6}\n{"-"*20}")
print(f"Dividing Two Vectors:\n{repr(v5)} / {repr(v6)}: {v5 / v6}\n{"-"*20}")
print(f"Comparing Two Vectors:\n{repr(v5)} == {repr(v6)}: {v5 == v6}\n{"-"*20}")
print(f"Less Than Condition:\n{repr(v5)} < {repr(v6)}: {v5 < v6}\n{"-"*20}")
print(f"Greater Than Condition:\n{repr(v5)} > {repr(v6)}: {v5 > v6}\n{"-"*20}")
print(f"Representation of First Vector:\n{repr(v5)}\n{"-"*20}")
print(f"Representation of second Vector:\n{repr(v6)}\n{"-"*20}")
```

## Type Validation with Custom Exceptions

Raising a custom `ErrorOccured` exception from inside `__init__` when the coordinates passed to `Vector` are not numbers, and catching it gracefully.

```python
# let say if we keep the different types or instances then??
class ErrorOccured(Exception):
    pass

class Vector:
    def __init__(self, x, y):
        # Ensure both inputs are either integers or floats
        if not isinstance(x, (int, float)) or not isinstance(y, (int, float)):
            raise ErrorOccured("Vector coordinates must be integers or floats!")
        
        self.x = x
        self.y = y
            
    def __add__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x + other.x, self.y + other.y)
        else:
            return NotImplemented
            
    def __sub__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x - other.x, self.y - other.y)
        else:
            return NotImplemented
            
    def __mul__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x * other.x, self.y * other.y)
        else:
            return NotImplemented
            
    def __truediv__(self,other):
        if isinstance(other,Vector):
            if other.x!=0 and other.y !=0:
                x_part=self.x / other.x
                y_part=self.y / other.y
                return Vector(x_part, y_part)
            else:
                print("can't divide as one of the vectors y has 0 value")
        else:
            return NotImplemented
               
    def __eq__(self,other):
        if isinstance(other,Vector):
            return self.x == other.x and self.y == other.y
        else:
            return False
            
    def __lt__(self,other):
        if isinstance(other,Vector):
            return self.x < other.x and self.y < other.y
        else:
            return False
            
    def __gt__(self,other):
        if isinstance(other,Vector):
            return self.x > other.x and self.y > other.y
        else:
            return False
            
    def __repr__(self):
        return f"Vector({self.x},{self.y})"
    
try:
    print("Attempting to create v5 with invalid coordinates (2, '4')...")
    v5 = Vector(2, '4')
    
except ErrorOccured as e:
    # This block intercepts the custom error and prints  exact message
    print(f"\n[ERROR CAUGHT]: {e}")
    print("-" * 30)
    
    # Fallback to a valid object so the remaining print scripts don't break
    print("Falling back to valid coordinates for v5: (2, 4)\n")
    v5 = Vector(2, 4)
    
v6 = Vector(5, 7)
print(f"Adding Two Vectors:\n{repr(v5)} + {repr(v5)}: {v5 + v6}\n{"-"*20}")
print(f"Substracting Two Vectors:\n{repr(v5)} - {repr(v6)}: {v5 - v6}\n{"-"*20}")
print(f"Multiplying Two Vectors:\n{repr(v5)} * {repr(v6)}: {v5 * v6}\n{"-"*20}")
print(f"Dividing Two Vectors:\n{repr(v5)} / {repr(v6)}: {v5 / v6}\n{"-"*20}")
print(f"Comparing Two Vectors:\n{repr(v5)} == {repr(v6)}: {v5 == v6}\n{"-"*20}")
print(f"Less Than Condition:\n{repr(v5)} < {repr(v6)}: {v5 < v6}\n{"-"*20}")
print(f"Greater Than Condition:\n{repr(v5)} > {repr(v6)}: {v5 > v6}\n{"-"*20}")
print(f"Representation of First Vector:\n{repr(v5)}\n{"-"*20}")
print(f"Representation of second Vector:\n{repr(v6)}\n{"-"*20}")
```

### Birth Year Calculation Example

A combined example: inheritance (`Birth` → `Birthyear_calculation`) plus a custom exception (`doberror`) that validates age, and a `__str__` override for clean output.

```python
class Error(Exception):
    pass

class doberror(Error):
    pass

class Birth:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
class Birthyear_calculation(Birth):
    
        def __init__(self,name,age,occupation):
            super().__init__(name,age)
            self.occupation=occupation
            
            try:
                if self.age<0:
                    raise doberror("Birth year should not be negative or 0")
                
            except doberror as e:
                print(f"[Error]: {e}")
                
        def __str__(self):
            if self.age>0:
                birth_year = 2026 - self.age
                return f"{self.name} is a {self.occupation} and was born in {birth_year}."
            return f"{self.name} (Invalid Birth Object)"  
            
# making object
obj=Birthyear_calculation("Muhammad Noman",20,"CS Student")
print(obj)
```
