## Inheritance In Python

Inheritance is a fundamental concept in Object-Oriented Programming (OOP) that allows a class to inherit attributes and methods from another class. This lesson covers single inheritance and multiple inheritance, demonstrating how to create and use them in Python.

```python
# let try to create base class Car
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype
        
    def derive(self):
        print(f"Person derives the {self.enginetype} car")
        
# now let make the objects
        
obj1=Car(4,4,'Petrol')
obj1.derive()
```

```python
# now try to inherit a new class from the base class
class Tesla(Car):
    def __init__(self,windows,doors,enginetype,is_selfdriving):
        super().__init__(windows,doors,enginetype) # that shows we re inheriting from the base class Car
        self.is_selfdriving=is_selfdriving       # this one is the additional feature of the tesla car
    
    def type(self):
        print(f"Is Tesla a Self Deriving Car?:{self.is_selfdriving}")
    
# now cretae the object of the derived class Tesla
tesla_obj=Tesla(4,4,"Electic",True)
tesla_obj.type()
```

## Polymorphism
Polymorphism is a core concept in Object-Oriented Programming (OOP) that allows objects of different classes to be treated as objects of a common superclass. It provides a way to perform a single action in different forms. Polymorphism is typically achieved through method overriding and interfaces

## Method Overriding
Imagine a parent class called Animal that has a method named makeSound(). By default, the animal just makes a generic noise.

If you create a child class called Dog, it inherits makeSound(). However, a generic noise doesn't fit a dog. So, the Dog class overrides the makeSound() method to print "Woof!". Similarly, a Cat class would override it to print "Meow!".

```python
# let take an example of the method overriding
#base class 
class Animal:
    def __init__(self,name):
        self.name=name
        
    def make_sound(self):
        return "A specific sound of"
    
# Derived class that will override the make_sound function
class Dog(Animal):
    
    def __init__(self,name):
        super().__init__(name)
        
    def make_sound(self):
        return f"{super().make_sound()} {self.name} is: woof"
     
# Derived class that will override the make_sound function   
class Cat(Animal):
    def __init__(self,name):
        super().__init__(name)
        
    def make_sound(self):
        return f"{super().make_sound()} {self.name} is: Meow"
    
# making a function that demonstrate ploymorphisim

def animal_speak(animal):
    return animal.make_sound()
  
# making objects  
dg=Dog('lucy')
ct=Cat('Mano')

print(dg.make_sound())
print(f"{ct.make_sound()}\n")
print("Now Ploymorphisim function output:\n")
print(animal_speak(dg))
print(animal_speak(ct))
```

```python
# let take another example
class Shape:
    def  __init__(self,shape_name):
        self.shape_name=shape_name
        
    def area(self):
        return f"The area of the"
    
class Rectangle(Shape):
    def __init__(self,shape_name,length,width):
        super().__init__(shape_name)
        self.length=length
        self.width=width
        
    def area(self):
        return f"{super().area()} {self.shape_name} is:{self.length * self.width}"
    
class Square(Shape):
    def __init__(self,shape_name,radius):
        super().__init__(shape_name)
        self.radius=radius
        
    def area(self):
        return f"{super().area()} {self.shape_name} is:{3.1415 * self.radius**2}"
    
# Now polyporphisim
def shape_area(parameter):
    print(f"{parameter.area()}")

# now let make objects
rec=Rectangle("Rectangle",4,5)
sqr=Square("Square",6)

# now call th polymorphisim finction
shape_area(rec)
shape_area(sqr)
```



## Polymorphism with Abstract Base Classes

Abstract Base Classes (ABCs) are used to define common methods for a group of related objects. They can enforce that derived classes implement particular methods, promoting consistency across different implementations.

```python
# Now let take another example by using the ABC
from abc import ABC,abstractmethod
# Base Class
class Vehical(ABC):
    @abstractmethod
    def __init__(self,vehical_name):
        self.vehical_name=vehical_name
        
    def engine_start(self):
        return f"Engine has been started"
    
# derived class
class Car(Vehical):
    def __init__(self,vehical_name):
        super().__init__(vehical_name)
        
    def engine_start(self):
        return f"{self.vehical_name} {super().engine_start()}"

class Motorcycle(Vehical):
    def __init__(self,vehical_name):
        super().__init__(vehical_name)
        
    def engine_start(self):
        return f"{self.vehical_name} {super().engine_start()}"
    
# Now Polymorphisim function
def engine_information(param):
    print(f"{param.engine_start()}")
    
# now let make the objet
cr=Car("BMW")
mtr=Motorcycle("Honda Prider")
# now call the function of the polymorphisim
engine_information(cr)
engine_information(mtr)
```

## Encapsulation And Abstraction

Encapsulation and abstraction are two fundamental principles of Object-Oriented Programming (OOP) that help in designing robust, maintainable, and reusable code. Encapsulation involves bundling data and methods that operate on the data within a single unit, while abstraction involves hiding complex implementation details and exposing only the necessary features.

### Encapsulation

Encapsulation is the concept of wrapping data (variables) and methods (functions) together as a single unit. It restricts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data.

```python
# let take first public attributes example
class Person:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender=gender
        
def get_values(param):
    return f"Name:{param.name}\nAge:{param.age}\nGender:{param.gender}"
# now object
prsn=Person("Muhammad Noman",22,"Male")
print(get_values(prsn))
```

```python
# now let take an example of the private attributes
# now try to use the getter and setter 
class Person:
    def __init__(self,name,age):
        self._name=name # private name
        self._age=age   # provate age
     # name getter   
    def get_name(self):
        return self._name
    # name setter
    def set_name(self,name):
       self._name=name
       
    # age getter 
    def get_age(self):
        return self._age
    # age setter
    def set_age(self,age):
        if age>0:
            self._age=age
        else:
            print("Age can't be negative")
            
class Employee(Person):
    def __init__(self,name,age):
        super().__init__(name,age)

        
    def get_name(self):
        return super().get_name()
    
    def set_name(self, name):
        return super().set_name(name)
    
    def get_age(self):
        return super().get_age()
    
    def set_age(self, age):
        return super().set_age(age)
    
# applying polymorphisim concept 
def getter(param):
    print(param.get_name())
    print(param.get_age())
    
# making object
emp=Employee("Muhammad Naseem",25)

print(emp.get_name())
print(emp.get_age())
print()
# using polymorphisim function on the protected variables
getter(emp)
```


**Abstraction**

Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object. This helps in reducing programming complexity and effort.

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    def drive(self):
        print("This is the drive function of the base class")
        
    @abstractmethod
    def engine_start(self): # just define 
        pass
    
class Car(Vehicle):
    def engine_start(self): # expand here and that is abstraction
        print("The engine has been started")
        
def car_info(param):
    param.engine_start()
    param.drive()

obj=Car()
car_info(obj) 
```

```python
from abc import ABC ,abstractmethod

class PaymentProcessor(ABC):
    def __init__(self,name):
        self._name=name
        
    def receipt(self,amount):
        print(f"Receipt generated for the amount ${amount}")
        
    @abstractmethod
    def process_payment(self,amount):
        pass
    
# now here abstract fuction will be open 
 
class CreditCardPayment(PaymentProcessor):
    def __init__(self,name):
        super().__init__(name)
        
    def process_payment(self, amount):
        print(f"{self._name} Securely processing ${amount} via Credit Card Gateway...")
        
class PayPalPayment(PaymentProcessor):
    def __init__(self,name):
        super().__init__(name)
        
    def process_payment(self, amount):
        print(f"{self._name} Securely processing ${amount} via Pay Pal Gateway...")
        
# polymorphisim function applying
def info(param,amount):
    param.process_payment(amount)
    param.receipt(amount)
    

card_obj=CreditCardPayment("Muhammad")
paypal=PayPalPayment("Muhammad")
info(card_obj,2500)
print("-" *20)
info(paypal,200)
```
