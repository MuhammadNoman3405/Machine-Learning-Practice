## Common Operator Overloading Magic Methods — Overview

A quick reference of the most common overloadable operators before diving into a full example.

```python
#### Common Operator Overloading Magic Methods
'''

__add__(self, other): Adds two objects using the + operator.
__sub__(self, other): Subtracts two objects using the - operator.
__mul__(self, other): Multiplies two objects using the * operator.
__truediv__(self, other): Divides two objects using the / operator.
__eq__(self, other): Checks if two objects are equal using the == operator.
__lt__(self, other): Checks if one object is less than another using the < operator.

'''
```

## Building a Vector Class with Full Operator Overloading

A complete `Vector` class implementing `+`, `-`, `*`, `/`, `==`, `<`, `>`, and `repr()` — the core toolkit of operator overloading in one example.

```python
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
        
    def __add__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x + other.x, self.y + other.y)
        else:
            return NotImplemented
            
    def __sub__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x - other.x, self.y - other.y)
        else:
            return NotImplemented
            
    def __mul__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x * other.x, self.y * other.y)
        else:
            return NotImplemented
            
    def __truediv__(self,other):
        if isinstance(other,Vector):
            if other.x!=0 and other.y !=0:
                x_part=self.x / other.x
                y_part=self.y / other.y
                return Vector(x_part, y_part)
            else:
                print("can't divide as one of the vectors y has 0 value")
        else:
            return NotImplemented
               
    def __eq__(self,other):
        if isinstance(other,Vector):
            return self.x == other.x and self.y == other.y
        else:
            return False
            
    def __lt__(self,other):
        if isinstance(other,Vector):
            return self.x < other.x and self.y < other.y
        else:
            return False
            
    def __gt__(self,other):
        if isinstance(other,Vector):
            return self.x > other.x and self.y > other.y
        else:
            return False
            
    def __repr__(self):
        return f"Vector({self.x},{self.y})"
    
 # making objects and performing tasks   
v5=Vector(2,4)
v6=Vector(6,8)
print(f"Adding Two Vectors:\n{repr(v5)} + {repr(v6)}: {v5 + v6}\n{"-"*20}")
print(f"Substracting Two Vectors:\n{repr(v5)} - {repr(v6)}: {v5 - v6}\n{"-"*20}")
print(f"Multiplying Two Vectors:\n{repr(v5)} * {repr(v6)}: {v5 * v6}\n{"-"*20}")
print(f"Dividing Two Vectors:\n{repr(v5)} / {repr(v6)}: {v5 / v6}\n{"-"*20}")
print(f"Comparing Two Vectors:\n{repr(v5)} == {repr(v6)}: {v5 == v6}\n{"-"*20}")
print(f"Less Than Condition:\n{repr(v5)} < {repr(v6)}: {v5 < v6}\n{"-"*20}")
print(f"Greater Than Condition:\n{repr(v5)} > {repr(v6)}: {v5 > v6}\n{"-"*20}")
print(f"Representation of First Vector:\n{repr(v5)}\n{"-"*20}")
print(f"Representation of second Vector:\n{repr(v6)}\n{"-"*20}")
```

## Type Validation with Custom Exceptions

Raising a custom `ErrorOccured` exception from inside `__init__` when the coordinates passed to `Vector` are not numbers, and catching it gracefully.

```python
# let say if we keep the different types or instances then??
class ErrorOccured(Exception):
    pass

class Vector:
    def __init__(self, x, y):
        # Ensure both inputs are either integers or floats
        if not isinstance(x, (int, float)) or not isinstance(y, (int, float)):
            raise ErrorOccured("Vector coordinates must be integers or floats!")
        
        self.x = x
        self.y = y
            
    def __add__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x + other.x, self.y + other.y)
        else:
            return NotImplemented
            
    def __sub__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x - other.x, self.y - other.y)
        else:
            return NotImplemented
            
    def __mul__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x * other.x, self.y * other.y)
        else:
            return NotImplemented
            
    def __truediv__(self,other):
        if isinstance(other,Vector):
            if other.x!=0 and other.y !=0:
                x_part=self.x / other.x
                y_part=self.y / other.y
                return Vector(x_part, y_part)
            else:
                print("can't divide as one of the vectors y has 0 value")
        else:
            return NotImplemented
               
    def __eq__(self,other):
        if isinstance(other,Vector):
            return self.x == other.x and self.y == other.y
        else:
            return False
            
    def __lt__(self,other):
        if isinstance(other,Vector):
            return self.x < other.x and self.y < other.y
        else:
            return False
            
    def __gt__(self,other):
        if isinstance(other,Vector):
            return self.x > other.x and self.y > other.y
        else:
            return False
            
    def __repr__(self):
        return f"Vector({self.x},{self.y})"
    
try:
    print("Attempting to create v5 with invalid coordinates (2, '4')...")
    v5 = Vector(2, '4')
    
except ErrorOccured as e:
    # This block intercepts the custom error and prints  exact message
    print(f"\n[ERROR CAUGHT]: {e}")
    print("-" * 30)
    
    # Fallback to a valid object so the remaining print scripts don't break
    print("Falling back to valid coordinates for v5: (2, 4)\n")
    v5 = Vector(2, 4)
    
v6 = Vector(5, 7)
print(f"Adding Two Vectors:\n{repr(v5)} + {repr(v5)}: {v5 + v6}\n{"-"*20}")
print(f"Substracting Two Vectors:\n{repr(v5)} - {repr(v6)}: {v5 - v6}\n{"-"*20}")
print(f"Multiplying Two Vectors:\n{repr(v5)} * {repr(v6)}: {v5 * v6}\n{"-"*20}")
print(f"Dividing Two Vectors:\n{repr(v5)} / {repr(v6)}: {v5 / v6}\n{"-"*20}")
print(f"Comparing Two Vectors:\n{repr(v5)} == {repr(v6)}: {v5 == v6}\n{"-"*20}")
print(f"Less Than Condition:\n{repr(v5)} < {repr(v6)}: {v5 < v6}\n{"-"*20}")
print(f"Greater Than Condition:\n{repr(v5)} > {repr(v6)}: {v5 > v6}\n{"-"*20}")
print(f"Representation of First Vector:\n{repr(v5)}\n{"-"*20}")
print(f"Representation of second Vector:\n{repr(v6)}\n{"-"*20}")
```

### Birth Year Calculation Example

A combined example: inheritance (`Birth` → `Birthyear_calculation`) plus a custom exception (`doberror`) that validates age, and a `__str__` override for clean output.

```python
class Error(Exception):
    pass

class doberror(Error):
    pass

class Birth:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
class Birthyear_calculation(Birth):
    
        def __init__(self,name,age,occupation):
            super().__init__(name,age)
            self.occupation=occupation
            
            try:
                if self.age<0:
                    raise doberror("Birth year should not be negative or 0")
                
            except doberror as e:
                print(f"[Error]: {e}")
                
        def __str__(self):
            if self.age>0:
                birth_year = 2026 - self.age
                return f"{self.name} is a {self.occupation} and was born in {birth_year}."
            return f"{self.name} (Invalid Birth Object)"  
            
# making object
obj=Birthyear_calculation("Muhammad Noman",20,"CS Student")
print(obj)
```

```python
obj2=Birthyear_calculation("Salman",-5," Student")
print(obj2)
```
