## Reading & Writing JSON

Loading data from a JSON file with `read_json()`, and converting a DataFrame back into JSON using different layouts — `orient='index'` and `orient='records'` — with `to_json()`.

```python
df=pd.read_json("students.json")
print(df)
```

```python
# again to json convert
df.to_json()
```

```python
# using the index orient
df.to_json(orient='index')
# that is giving the values index wise
```

```python
# shows the original one given data 
df.to_json(orient='records')
```

## Reading CSV from a URL

Reading a CSV file directly from a URL — no need to download it first — then saving the result back to a local file with `to_csv()`.

```python
df=pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data")
```

```python
df.head(5)
```

```python
df.size
```

```python
df.to_csv('wine.csv')
```

<a id="reading-html"></a>
## Reading Tables from HTML (Web Scraping)

Pulling every table straight out of a webpage with `read_html()` — it returns a list of all tables found on the page, so you pick out the one you actually want.

```python
url='https://www.fdic.gov/bank-failures/failed-bank-list/'
```

```python
df=pd.read_html(url)
```

```python
df=df[0]
df
```

```python
df.set_index('Bank Name', inplace=True)
```

```python
df
```
