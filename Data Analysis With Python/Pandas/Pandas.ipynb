````python
points_df=pd.DataFrame(
    {
        'points_01':[(3,4), (3,6), (-2,4), (-5,4), (0,4), (-4,4), (2,4)],
        'points_02':[(6,4), (-6,-6), (5,2), (-9,4), (1,4), (8,-9), (3,8)]
    }
)
points_df
````

````python
def euclidean_distance(row):
    point_a=row['points_01']
    point_b=row['points_02']
    return ((point_a[0]- point_b[0])**2 + (point_a[1]- point_b[1])**2)**0.5
````

````python
points_df.apply(euclidean_distance,axis=1)
# that is the th aapply function 
# the value will 
````
