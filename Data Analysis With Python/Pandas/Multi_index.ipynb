```python
import numpy as np
import pandas as pd
```

```python

# how to create multiindex object
# 1. pd.MultiIndex.from_tuples()
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
multi_index=pd.MultiIndex.from_tuples(index_val)
multi_index
# 2. pd.MultiIndex.from_product()
```

```python
# 2. can also make by the cartesian product
multi_index_2=pd.MultiIndex.from_product([['cse','ece'],[2021,2022,2023,2024]])
multi_index_2
```

```python
# gives the levels
multi_index.levels
```

```python
multi_index_2.levels
```

```python
# fetching th levels
multi_index.levels[0]
```

```python
multi_index_2.levels[0]
```

```python
multi_index.levels[1]
```

```python
multi_index_2.levels[1]
```

```python
# creating a series from the  multi_index objects
s=pd.Series([1,2,3,4,5,6,7,8],index=multi_index)
s
```

```python
# now fetching the elements
s['cse']
```

```python
print(s[('cse',2021)])
```

```python
# we can also make its 2d dfa viva unstack()
temp=s.unstack()
temp
```

```python
temp.stack()
```
