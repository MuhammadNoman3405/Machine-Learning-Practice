How arithmetic and comparisons work on NumPy arrays — scalar ops, relational ops, and vector-to-vector ops.

### Scalar Operations

Applying a single number to an entire array at once — NumPy broadcasts the operation across every element.

````python
# Scalar OPerations

arry=np.arange(12)
print(arry) # original one
new_one=arry * 2
print(new_one) # updated 
````

````python
var=arry ** 2 # takes the square of the entire array
print(var)
````

<a id="np-array-functions"></a>
#### Array Functions

Built-in functions that summarize or transform arrays — sums, stats, trig, dot products, and rounding.

<a id="np-max-min-sum-prod"></a>
### Max, Min, Sum, Product

Aggregate functions that reduce an array down to a single summary value — or per-axis values.

````python
A1=np.random.random((4,4)) # 4 rows and 4 columns
round=np.round(A1 * 100)
print(round)
````

````python
# max/min/sum/prod
max=np.max(round)
min=np.min(round)
print(max)
print(min)
print(np.sum(round)) # gives the sum of the all items
print(np.prod(round)) # gives the product of the all elements
````

````python
#  for axis row=1 and col=0 
print(f"Original Matrix:\n {round}\n")

column_wise_max=np.max(round, axis=0) # axis=0 mean column wise max
row_wise_max=np.max(round, axis=1) # axis=1 mean row wise max
print(f"Column wise max:\n{column_wise_max}\n")
print(f"Row wise max:\n {row_wise_max}")
````

<a id="np-stats-functions"></a>
#### Mean/Median/Std/Var

`np.mean()`, `np.median()`, `np.std()`, and `np.var()` — core statistical summaries, usable row-wise or column-wise via `axis`.

````python
import numpy as np
mean=np.mean(round,axis=1) # row wise mean
median=np.median(round,axis=1)
standard_deviation=np.std(round,axis=0) # column wise std
varience=np.var(round,axis=0) # column wise varience is taken 
print(mean)
print(median)
print(standard_deviation)
print(varience)
````

### Speed Comparison

Benchmarking a Python list comprehension against a NumPy array for the same operation.

````python
# speed
a=[i for i in range(10000000)]
b=[i for i in range(10000000,20000000)]

c=[]

import time 
start = time.time()
for i in range(len(a)):
    c.append(a[i] + b[i])
end=time.time() - start
print(end)
````

````python
import numpy as np
a=np.arange(10000000)
b=np.arange(10000000,20000000)  # time taken by the numpy
start=time.time()
add=a+b
end=time.time() - start
print(end)
````

## Broadcasting

"The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations.

The smaller array is "broadcast" across the larger array so that they have compatible shapes.

## Different Shape

What happens when NumPy has to stretch a smaller array to match a bigger one during an operation.

````python
# Take the different shapes two array
a=np.arange(6).reshape(2,3)
b=np.arange(3).reshape(1,3)
print(a)
print()
print(b)
#  now let try to add them
print(f"Result of Addition:\n{a+b}")
````

<a id="np-broadcasting-rules"></a>
## Broadcasting Rules

The two rules NumPy follows to decide whether two differently-shaped arrays can be combined in an operation.

## 1 Make the two arrays have the same number of dimensions.

If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head of the array with the smaller dimension.

## 2 Make each dimension of the two arrays the same size.

If the sizes of each dimension of the two arrays do not match, dimensions with size 1 are stretched to the size of the other array.  

If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted, and an error is raised.

A common regression loss function — the average of the squared differences between actual and predicted values.

<a id="np-cce"></a>
### Categorical Cross-Entropy (CCE)

A loss function for multi-class classification, comparing predicted probability distributions against true one-hot labels.

````python
def cce(y_true,y_pred):
    y_pred = np.clip(y_pred, 1e-15, 1.0 - 1e-15) # prevents from infinity
    loss=-np.sum(y_true * np.log(y_pred))
    return np.mean(loss)

y_true=np.array([
    [1, 0, 0],  # Sample 1 is a Cat
    [0, 1, 0],  # Sample 2 is a Dog
    [0, 0, 1]   # Sample 3 is a Bird
])
y_pred=np.array([
    [0.80, 0.15, 0.05], # Pretty good prediction for Cat
    [0.10, 0.70, 0.20], # Decent prediction for Dog
    [0.20, 0.20, 0.60]  # Confused, but got Bird right
])

cce_loss=cce(y_true,y_pred)
print(f"Categorial Crosss Entropy: {cce_loss:.4f}")
````

## Split, Apply, and Combine

The core pattern behind GroupBy: split the data into groups, apply a custom function to each group, then combine the results back together — done here with `.apply()` and custom functions.

## 1 Relational Plot

Using `sns.relplot()` to explore relationships between two numeric variables, with `hue`, `col`, `row`, `size`, and `palette` to bring in extra dimensions.

````python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# relpot
tips=sns.load_dataset('tips')
sns.relplot(data=tips,x=tips['total_bill'],y=tips['tip'],hue='day')
plt.show()
````

````python
# adding the columns
sns.relplot(data=tips,x=tips['total_bill'],y=tips['tip'],hue='day',col='time')
````

````python
# adding row
sns.relplot(data=tips,x=tips['total_bill'],y=tips['tip'],hue='day',col='time',row='sex')
````

````python
# When the variable assigned to col has many levels, it can be
# “wrapped” across multiple rows:
sns.relplot(data=tips,x=tips['total_bill'],y=tips['tip'],hue='time',col='day',col_wrap=2)
plt.show()
````

````python
sns.relplot(data=tips,x=tips['total_bill'],y=tips['tip'],hue='time',col='time',size='size',palette=['b','r'],style='sex',sizes=(10, 100))
plt.show()
````

<a id="sb-scatter-plot-seaborn"></a>
## 02 Scatter Plot

A deep dive into `sns.scatterplot()` — mapping color (`hue`), shape (`style`), and size (`size`) to extra variables, and controlling palettes, marker styles, and legends.

````python
sns.scatterplot(x='total_bill',y='tip',data=tips,hue='day',palette=['b','r'])
plt.show()
````

````python
# hue on the basis on time
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time")
plt.show()
````

````python
# style
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='time',style='time')
plt.show()
````

````python
# hue and style
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='day',style='time')
plt.show()
````

````python
# same can be done
sns.relplot(data=tips,x='total_bill',y='tip',hue='day',style='time')
plt.show()
````

````python
# If the variable assigned to hue is numeric, the semantic
# mapping will be quantitative and use a different default palette:
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='size')
plt.show()
````

````python
# Pass the name of a categorical palette or explicit colors
# (as a Python list of dictionary) to force categorical mapping 
# of the hue variable:
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='size',palette='deep')
plt.show()
````

````python
# If there are a large number of unique numeric values, the legend 
# will show a representative, evenly-spaced set:
tip_rate=tips.eval("tip / total_bill").rename('tip_rate')
sns.scatterplot(x='total_bill',y='tip',hue=tip_rate,data=tips)
````

````python
sns.scatterplot(x='total_bill',y='tip',hue='size',size='size',palette='deep',style='size',data=tips)
plt.show()
````

````python
# Control the range of marker areas with sizes, and set
# legend="full" to force every unique value to appear in the legend:
sns.scatterplot(x='total_bill',y='tip',hue='size',size='size',sizes=(20,200),legend="full",palette='deep',style='size',data=tips)
plt.show()
````

````python
# Pass a tuple of values or a matplotlib.colors.
# Normalize object to hue_norm to control the quantitative hue mapping:
sns.scatterplot(x='total_bill',y='tip',hue='size',size='size',sizes=(20,200),legend="full",palette='deep',hue_norm=(0,7),data=tips)
plt.show()
````

````python
# Control the specific markers used to map the style variable
# by passing a Python list or dictionary of marker codes:
markers = {"Lunch": "d", "Dinner": "X"}
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time",style="time", markers=markers)
````

````python
sns.scatterplot(data=tips, x="total_bill", y="tip", s=100, color=".1", marker="+")
````

````python
sns.relplot(
    data=tips, x="total_bill", y="tip",
    col='time',hue='day',style='day',size='size'
)
````

<a id="sb-displot"></a>
## 01 Displot

Using `sns.displot()` as a flexible figure-level function for distributions — switching between histogram, KDE, and ECDF views, bivariate plots, and splitting by category with `hue`, `col`, and `multiple='stack'`.

````python
penguins=sns.load_dataset('penguins')
penguins.head(10)
````

````python
sns.displot(data=penguins,x='flipper_length_mm')
````

````python
# give the kind kernel density estimates (KDEs),
sns.displot(data=penguins,x='flipper_length_mm',kind='kde',fill='full')
````

````python
# empirical cumulative distribution functions (ECDFs):
sns.displot(data=penguins,x='flipper_length_mm',kind='ecdf')
````

````python
# kind is histogram
sns.displot(data=penguins,x='flipper_length_mm',kind='hist')
````

````python
sns.displot(data=penguins,x='flipper_length_mm',kde=True)
````

````python
sns.displot(data=penguins,x='flipper_length_mm',y='bill_length_mm')
````

````python
# For each kind of plot, you can also show individual observations
# with a marginal “rug”:
sns.displot(data=penguins,x='flipper_length_mm',y='bill_length_mm',
            kind='kde',rug=True,color='crimson')
````

````python
# Each kind of plot can be drawn separately for subsets of data 
# using hue mapping:
sns.displot(data=penguins,x='flipper_length_mm',hue='species',kind='kde')
````

````python
sns.displot(data=penguins,x="flipper_length_mm",hue='sex',
            col='species',kde=True)
````

<a id="sb-histplot"></a>
## 02 Histogram

Using `sns.histplot()` directly — controlling bin width and bin count, adding a KDE overlay, plotting on the y-axis, and drawing wide-form histograms for every numeric column at once.

````python
sns.histplot(data=penguins,x="flipper_length_mm",hue='species',multiple="stack")
````

````python
sns.histplot(data=penguins,y="flipper_length_mm",hue='species',multiple="stack")
````

````python
# Check how well the histogram represents the data by specifying
# a different bin width:
sns.histplot(data=penguins,x="flipper_length_mm",hue="species",multiple="stack",binwidth=3)
````

````python
# You can also define the total number of bins to use
plt.figure(figsize=(9,5))
sns.histplot(data=penguins,x="flipper_length_mm",hue="species",multiple="stack",bins=30)
````

````python
# adding kde
plt.figure(figsize=(9,5))
sns.histplot(data=penguins,x="flipper_length_mm",hue="species",multiple="stack",bins=30,kde=True)
````

````python
# If neither x nor y is assigned, the dataset is treated
# as wide-form, and a histogram is drawn for each numeric column:
sns.histplot(data=penguins)
````

````python
# Overlapping bars can be hard to visually resolve. A different 
# approach would be to draw a step function:
sns.histplot(data=penguins,x="flipper_length_mm",hue="species",multiple="stack",element="step")
````

<a id="sb-histplot-poly-note"></a>
You can move even farther away from bars by drawing a polygon with vertices in the center of each bin. This may make it easier to see the shape of the distribution, but use with caution: it will be less obvious to your audience that they are looking at a histogram:

````python
sns.histplot(data=penguins,x="flipper_length_mm",hue="species",multiple="stack",element="poly")
plt.show()
````
