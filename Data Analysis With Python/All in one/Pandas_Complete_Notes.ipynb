<a id="s-series-important-methods"></a>
## Some Important Series Methods

A deeper toolbox of Series methods — `agg`, `astype`, `between`, `clip`, `drop_duplicates`, `dropna`, `fillna`, `filter`, `iloc`, `isnull`, and `map`.

````python
import numpy as np
import pandas as pd

# 1. Setup a sample Series to practice on
prices = pd.Series(
    [100, 250, 100, np.nan, 50, 1200, 300],
    index=["item1", "item2", "item3", "item4", "item5", "item6", "item7"],name='Prices List'
)
````

````python
print("Original Series:")
print(prices)
print("-" * 40)
````

````python
# 2. agg (Aggregate one or more operations)
# Finds multiple summary statistics at once

print("Agg:")
print(prices.agg(["min", "max", "mean"]))
print("-" * 40)
````

````python
# 3. astype (Change data type)
# Converts data type (e.g., changing float to string or integer)
# Note: We drop NaN first because integers can't hold standard NaNs
print("Astype:")
print(prices.dropna().astype(int))
print("-" * 40)
````

````python
print("Between")
print(prices.between(100,300))
print('*' *40)
````

````python
# i can also do the fancy indexing on that
print("Between")
print(prices[prices.between(100,300)])
print('*' *40)
````

````python
prices
````

````python
# 5. clip (Cap outliers or extreme values)
# Forces values below 100 to become 100, and values above 500 to become 500
print("Clip")
print(prices.clip(lower=100,upper=500))
print("-"*40)
````

````python
# 6. drop_duplicates (Remove repeating values)
# Removes the second '100' in our dataset
print("Drop_duplicates:")
print(prices.drop_duplicates())
print("-" * 40)
````

````python
# 7. dropna (Remove missing values)
# Drops the NaN value at 'item4'
print("Dropna:")
print(prices.dropna())
print("-" * 40)
````

````python
prices
````

````python
# 8. fillna (Replace missing values)
# Replaces the NaN at 'item4' with a default value like 0
print(prices.fillna(0)) # see the item numner 04
````

````python
# 9. filter (Filter by index labels)
# Searches and keeps rows where the index label matches a pattern
print("Filter:")
print(prices.filter(like="item6"))
````

````python
# 10. iloc (Select by integer position)
# Grabs the value at index position 1 (the 2nd element
print("iloc")
print(prices.iloc[1])
````

````python
# 11. isnull (Check for missing values)
# Returns True if a value is NaN, False otherwise
print("isnull:")
print(prices.isnull())
print("-" * 40)
````

````python
# printing the exact null value through fancy indexing
print("isnull:")
print(prices[prices.isnull()]) 
print("-" * 40)
````

````python
# 12. map (Transform values using a dictionary or function)
# Modifies every value in the Series (e.g., adding a currency sign)
print("map:")
print(prices.map(lambda x: f"PKR {x}"))
# lambda argument : expression
````

````python
s = pd.Series(["cat", "dog", np.nan, "rabbit"])
s
````

````python
# map accepts a dict or a Series. Values that are not found 
# in the dict are converted to NaN,unless the dict has a defaultvalue (e.g. defaultdict):
s.map({"cat": "kitten", "dog": "puppy"})
````

````python
# It also accepts a function:
s.map("i am a {}".format)
````

````python
# It also accepts a function:
s.dropna().map("i am a {}".format,na_action="ignore")
````

````python
# It also accepts a function:
s.dropna().map("i am a {}".format)
````

The core pattern behind GroupBy: split the data into groups, apply a custom function to each group, then combine the results back together — done here with `.apply()` and custom functions.

````python
# here i am getting the movies names that are starting with the letter T
def foo(group):
    return group['Title'].str.startswith('T').sum()

genre.apply(foo)
````

````python
# let try to find out the every group rank on the apply function basis
def rank_movie(group):
    group['Rank_movies']=group['IMDB_Rating'].rank(ascending=False)
    return group
genre.apply(rank_movie)
````

<a id="gb-split-apply-combine-diagram"></a>


*Visual illustration of the split → apply → combine process shown above.*

````python
# find the normalized IMDB Rating
# this is the example of the split , apply then and combine
def normalized_rating(group):
    group['normalized_rating']=group['IMDB_Rating'] - group['IMDB_Rating'].min()/(group['IMDB_Rating'].max()-group['IMDB_Rating'].min())
    return group
genre.apply(normalized_rating)
````
