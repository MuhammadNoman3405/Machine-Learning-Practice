## Broadcasting

"The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations.

The smaller array is "broadcast" across the larger array so that they have compatible shapes.

<a id="broadcasting-same-shape"></a>

<a id="broadcasting-different-shape"></a>
## Different Shape

What happens when NumPy has to stretch a smaller array to match a bigger one during an operation.

```python
# Take the different shapes two array
a=np.arange(6).reshape(2,3)
b=np.arange(3).reshape(1,3)
print(a)
print()
print(b)
#  now let try to add them
print(f"Result of Addition:\n{a+b}")
```

<a id="broadcasting-rules"></a>
## Broadcasting Rules

The two rules NumPy follows to decide whether two differently-shaped arrays can be combined in an operation.

## 1 Make the two arrays have the same number of dimensions.

If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head of the array with the smaller dimension.

## 2 Make each dimension of the two arrays the same size.

If the sizes of each dimension of the two arrays do not match, dimensions with size 1 are stretched to the size of the other array.  

If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted, and an error is raised.

<a id="mse"></a>
### Mean Squared Error (MSE)

A common regression loss function — the average of the squared differences between actual and predicted values.

```python
def mse(actula,predict):
    return np.mean((actual - predict)**2)

actual=np.random.randint(1,100,30)
prediction=np.random.randint(1,100,30)
print(f"Actual Data:\n{actual}\n")
print(f"Predicted Data:\n{prediction}\n")
print(f"Mean Squared Error:\n{mse(actual,prediction)}")
```

<a id="cce"></a>
### Categorical Cross-Entropy (CCE)

A loss function for multi-class classification, comparing predicted probability distributions against true one-hot labels.

```python
def cce(y_true,y_pred):
    y_pred = np.clip(y_pred, 1e-15, 1.0 - 1e-15) # prevents from infinity
    loss=-np.sum(y_true * np.log(y_pred))
    return np.mean(loss)

y_true=np.array([
    [1, 0, 0],  # Sample 1 is a Cat
    [0, 1, 0],  # Sample 2 is a Dog
    [0, 0, 1]   # Sample 3 is a Bird
])
y_pred=np.array([
    [0.80, 0.15, 0.05], # Pretty good prediction for Cat
    [0.10, 0.70, 0.20], # Decent prediction for Dog
    [0.20, 0.20, 0.60]  # Confused, but got Bird right
])

cce_loss=cce(y_true,y_pred)
print(f"Categorial Crosss Entropy: {cce_loss:.4f}")
```
