## Setup — Connecting to SQLite

Importing `sqlite3`, opening a connection to a database file, and creating a cursor object to run SQL commands through.

```python
import sqlite3
```

```python
connection=sqlite3.connect("example.db")
connection
```

```python
# making the object of that
object=connection.cursor()
```

<a id="creating-table"></a>
## Creating a Table

Defining the `employees` table's structure with `CREATE TABLE IF NOT EXISTS` — columns, data types, and an auto-incrementing primary key.

```python
object.execute('''
               create table if Not Exists employees(
                   id integer not null primary key autoincrement,
                   name text,
                   age integer,
                   department text
               )
               ''')
connection.commit()
```

<a id="inserting-records"></a>
## Inserting Records

Adding rows to the table with `INSERT INTO ... VALUES`, then committing the changes so they're saved to the database file.

```python
object.execute(
    '''
    insert into employees(name,age,department)
    values('Muhammad Noman',22,'Computer Science')
    '''
)
object.execute(
    '''
    insert into employees(name,age,department)
    values('Muhammad Zeeshan',24,'Civil')
    '''
)
object.execute(
    '''
    insert into employees(name,age,department)
    values('Muhammad Salman',20,'Software Engineering')
    '''
)
connection.commit()
```

<a id="querying-records"></a>
## Querying Records (SELECT & WHERE)

Reading rows back out with `SELECT`, filtering with `WHERE` on a text column and a numeric comparison, and looping over `fetchall()` to print each row.

```python
object.execute('select * from employees where department="Civil"')
rows=object.fetchall()
for row in rows:
    print(row)
```

```python
object.execute('select * from employees where age>=22')
rows=object.fetchall()
for row in rows:
    print(row)
```

<a id="deleting-records"></a>
## Deleting Records

Removing a specific row with `DELETE FROM ... WHERE`, then querying again to confirm it's gone.

```python
object.execute('Delete  from employees where id=2 ')
```

```python
object.execute('select * from employees where age>=22')
rows=object.fetchall()
for row in rows:
    print(row)
```

```python
object.execute('select * from employees')
rows=object.fetchall()
for row in rows:
    print(row)
```

<a id="updating-records"></a>
## Updating Records

Changing an existing value with `UPDATE ... SET ... WHERE`, then querying again to see the updated row.

```python
# update the Salman age
object.execute('update employees set age=18 where id=3')
```

```python
object.execute('select * from employees ')
rows=object.fetchall()
for i in rows:
    print(i)
```

<a id="sales-data-examples"></a>
## Sales Data Examples

A second, bigger example — creating a fresh `sales` database and table, then bulk-inserting 10 rows at once with `executemany()`.

```python
conn=sqlite3.connect("sales_data.db")
```

```python
object=conn.cursor()
```

```python
object.execute('''
               CREATE TABLE IF NOT EXISTS sales (
    sale_id INTEGER PRIMARY KEY AUTOINCREMENT,
    product_name TEXT NOT NULL,
    category TEXT NOT NULL,
    quantity INTEGER NOT NULL,
    unit_price REAL NOT NULL,
    sale_date DATE NOT NULL,
    customer_region TEXT NOT NULL
)
               ''')
sales_data=[
    ("Laptop", "Electronics", 2, 850.00, "2026-07-01", "North"),
    ("Wireless Mouse", "Electronics", 5, 25.50, "2026-07-02", "East"),
    ("Office Chair", "Furniture", 1, 150.00, "2026-07-03", "West"),
    ("Coffee Mug", "Home & Kitchen", 10, 12.00, "2026-07-05", "South"),
    ("Mechanical Keyboard", "Electronics", 3, 75.00, "2026-07-10", "North"),
    ("Desk Lamp", "Furniture", 2, 45.00, "2026-07-12", "East"),
    ("Monitor", "Electronics", 1, 220.00, "2026-07-15", "West"),
    ("Notebook Pack", "Stationery", 8, 8.50, "2026-07-18", "South"),
    ("Ergonomic Mouse", "Electronics", 4, 35.00, "2026-07-20", "North"),
    ("Standing Desk", "Furniture", 1, 400.00, "2026-07-22", "South")
]
object.executemany(
    """
INSERT INTO sales (product_name, category, quantity, unit_price, sale_date, customer_region)
VALUES (?, ?, ?, ?, ?, ?)
""",sales_data,
)
```

```python
conn.commit()
```

```python
# now try to fetch the values
object.execute('select * from sales')
rows=object.fetchall()
for row in rows:
    print(row)
```

<a id="sql-queries"></a>
## Now try to do some queries

Writing more realistic SQL queries against the sales table — filtering by region, calculating revenue with an expression, aggregating with `GROUP BY`, and combining multiple conditions.

```python
# select all values where region is South
object.execute('select * from sales where  customer_region="South"')
values=object.fetchall()
for value in values:
    print(value)
```

```python
object.execute("""
               SELECT product_name, quantity, unit_price, (quantity * unit_price) AS total_revenue 
FROM sales;
               """)
values=object.fetchall()
for value in values:
    print(value)
```

```python
# calculate the total sales by category
object.execute("""
               select category,sum(quantity * unit_price) as Total_sales
               from sales
               group by category
               order by Total_sales desc
               """)
values=object.fetchall()
for value in values:
    print(value)
```

```python
# Filtering: Find all sales from the 'North' region where
# total spending was greater than $100.
# calculate the total sales by category
object.execute("""
               select * from sales 
               where  customer_region="North" and (quantity * unit_price)>100
               """)
values=object.fetchall()
for value in values:
    print(value)
```
