## Label Encoding

```python
import pandas as pd
df=pd.DataFrame({
    'Color':["red",'green','blue','red','green']
})
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder=LabelEncoder()
```

```python
df['labeled_column']=lbl_encoder.fit_transform(df['Color'])
print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})

le = LabelEncoder()
df['City_encoded'] = le.fit_transform(df['City'])

print(df)
```

## Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame({
    'size':['small','medium','large','small','large','large']
})
df
```

```python
ordinal_encoder=OrdinalEncoder(categories=[['small','medium','large']]) 
# assigning rank
ordinal_encoder.fit_transform(df[['size']])
```

```python
rank=ordinal_encoder.transform([['small']])
print(rank)
```

## One Hot Encoding

```python
import pandas as pd

df = pd.DataFrame({
    'Color': ['Red', 'Blue', 'Green', 'Red']
})

df_encoded = pd.get_dummies(df, columns=['Color'],dtype=int)
print(df_encoded)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# 1. Data banao
df = pd.DataFrame({
    'Color': ['Red', 'Blue', 'Green', 'Red', 'Blue']
})

# 2. Encoder banao
encoder = OneHotEncoder(sparse_output=False, drop=None)
# sparse_output=False ka matlab: direct array de do, matrix nahi

# 3. Fit + Transform karo
encoded = encoder.fit_transform(df[['Color']])

# 4. Columns ke naam lo
encoded_df = pd.DataFrame(
    encoded, 
    columns=encoder.get_feature_names_out(['Color'])
)

# 5. Original ke saath join karo
final_df = pd.concat([df, encoded_df], axis=1)

print(final_df)
```

## Target Guided Ordinal Encoding

```python
import pandas as pd

# create a sample dataframe with a categorical variable and a target variable
df = pd.DataFrame({
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'New York', 'Paris'],
    'price': [200, 150, 300, 250, 180, 320],
})
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
mean_price
```

```python
# maping the city value
df['city_encoded']=df['city'].map(mean_price)
df
```

```python
# shiw the difference 
df[['price','city_encoded']]
```
