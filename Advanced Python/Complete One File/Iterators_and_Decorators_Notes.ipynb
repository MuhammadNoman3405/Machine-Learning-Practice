# Part 1: Iterators & Generators

<a id="iter-intro"></a>

### Iterators

Iterators are advanced Python concepts that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

<a id="for-loop-iteration"></a>
## Iterating with a Simple For Loop

The most familiar way to loop through a collection — Python handles the iteration protocol automatically under the hood.

```python
# let take a simple for loop iterator
my_list=[1,2,3,4,5,6]
for items in my_list:
    print(items)
```

```python
print(type(my_list))
```

```python
my_list
```

<a id="iter-function"></a>
## Creating an Iterator Manually with `iter()`

Calling `iter()` on a list converts it into a `list_iterator` object — this is what powers every `for` loop behind the scenes.

```python
# now try to apply the advanced concept
iterator=iter(my_list)
print(type(iterator))
```

```python
iterator # checks its mem location 
```

<a id="next-function"></a>
## Accessing Elements with `next()`

`next()` pulls the next value out of an iterator, one at a time, advancing its internal position.

```python
# now how to access the elements of the list which  are stored in  the iterator
# for tht purpose wwe will use next()
next(iterator)
```

```python
next(iterator)
```

```python
next(iterator) # mean similarly it will just return me only next elements
```

```python
for x in iterator:
    print(x) # it will give us the remaing elements
```

<a id="stopiteration"></a>
## Exhausting an Iterator: StopIteration

Once every element has been consumed, calling `next()` again raises `StopIteration`. Wrapping it in a `try/except` handles this gracefully.

```python
next(iterator) # because we have access all the items of the iterator
```

```python
# to overcope with this problem try : block use
try:
    next(iterator)
    
except StopIteration:
    print("You have accessed all the elements no one is remaing to printout")
```

<a id="string-iterator"></a>
## Iterating Over Strings

Strings are also iterable — calling `iter()` on a string returns an iterator that yields one character at a time.

```python
# let say take a string and store it in a iterator var
my_string="Hello"
var=iter(my_string)
# will return one character at each call
print(next(var))
print(next(var))
print(next(var))
print(next(var))
print(next(var))
```

```python
# now if want to iterate again will give Stopiteration error
print(next(var))
```

<a id="generators-intro"></a>

**Generators**

Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

<a id="return-vs-yield"></a>
## Return vs Yield: The Difference

`return` exits a function immediately with a single value. `yield` pauses the function, remembers its state, and resumes exactly where it left off on the next call.

```python
# locally we see this concept 
def square(n):
    for i in range(n + 1):
        print(f"{i**2}")
    
square(3)
```

```python
# return only gives one value return
def square(n):
    for i in range(n + 1):
        return i**2
    
square(3)
```

<a id="creating-generator"></a>
## Creating a Generator with `yield`

Swapping `return` for `yield` turns an ordinary function into a generator function — calling it returns a generator object instead of running the code immediately.

```python
# if we use the keyword yield then it will become the generator
def square_value(n):
    for i in range(n + 1):
        yield i**2
    
square_value(3)
```

```python
# how to access all elements one by one  stored in the generator
for items in square_value(3):
    print(items)
```

```python
# i can also access it viva next method
def square_value2(n):
    for i in range(n + 1):
        yield i**2
    
square_value2(3)
```

<a id="generator-next"></a>
## Accessing Generator Values with `next()`

Generators support `next()` just like iterators do, pulling one value at a time.

```python
a=square_value2(3)
print(next(a))
print(next(a))
print(next(a))
print(next(a))
```

```python
# i can i also get viva applying  the for loop
# i can also access it viva next method
def square_value3(n):
    for i in range(n + 1):
        yield i**2
    
square_value3(3)
```

```python
var=square_value3(3)
for i in var:
    print(i)
```

```python
def generate():
    yield 1 # yield similarly works as return
    yield 2
    yield 3
    
var=generate()
for i in var:
    print(i)
```

<a id="generator-exhaustion"></a>
## Generator Exhaustion

Just like iterators, generators raise `StopIteration` once every value has been yielded — they can only be consumed once.

```python
# now it will throw an eror if i again want to print out its element
next(var)
```

```python
def generate1():
    yield 1 # yield similarly works as return
    yield 2
    yield 3
    
var=generate1()
var
```

```python
print(next(var))
print(next(var))
print(next(var))
```

<a id="generator-file-example"></a>
## Practical Example: Reading Files Lazily

A real-world use case — reading a large file line by line with a generator, so the whole file never needs to be loaded into memory at once.

```python
# try to practical example
def read_file(path):
    with open(path,'r') as file:
        for line in file:
            yield line
```

```python
file_path='large_file.txt'
for line in read_file(file_path):
    print(line.strip())
```

<a id="decorator-before-after"></a>
## Practical Example: Before/After Logic

A clean, classic decorator pattern: run code before the wrapped function, call it, then run code after — the foundation of logging, timing, and validation decorators.

```python
def decorator_fun(func):
    def wrapper():
        print("Something will happen before the function")
        func()
        print("After the function")
    return wrapper # this will return the object only to the main function 

@decorator_fun
def greeting():
    print("Hello Noman How Are You?")
greeting()  
```

```python
# let say return wrapper return output 
# mean return wrapper() will apply for the same function
def decorator_fun(func):
    def wrapper():
        print("Something will happen before the function")
        func()
        print("After the function")
    return wrapper() # this enforce to return the output of this finction

@decorator_fun
def greeting():
    print("Hello Noman How Are You?")
# no need to call the greetin function
```

<a id="decorator-with-arguments"></a>
## Decorators with Arguments

A decorator factory — `repeat(n)` returns a decorator that itself wraps a function, allowing the decorator to accept its own arguments like `n` (how many times to repeat).

```python
def repeat(n):
    def decorator(fun):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                fun(*args,**kwargs)
        return wrapper()
    return decorator

@repeat(5)
def greeting():
    print("Hy Noman")
```
