
**Decorators**

Decorators are a powerful and flexible feature in Python that allows you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

## 02 Closures

```python
def main_function():
    msg ="Hey Noman!"
    def sub_function():
        print(msg) # this is possible due to the closures concept
        print("Hello now you are learning the advance python")
        print("Hey Noman this is the Sub function of the Main function")
        
    return sub_function()
```

```python
main_function()
```

```python
# let try to give the parameter at he run time
def main_function(msg):
    def sub_function():
        print(msg) # this is possible due to the closures concept
        print("Hello now you are learning the advance python")
        print("Hey Noman this is the Sub function of the Main function")
        
    return sub_function()
```

```python
# we can also pass the inbuilt function also
def main_function(param):
    msg ="Hey Noman!"
    def sub_function():
        print(msg) # this is possible due to the closures concept
        param()
        print("Hello now you are learning the advance python")
        print("Hey Noman this is the Sub function of the Main function")
        
    return sub_function()
```

```python
main_function(print)
```

## Another Way 

```python

def main_function(func):
    def sub_function():
        inner=func("Ali",23)
        print(inner())
        print("Hello now you are learning the advance python")
        print("Hey Noman this is the Sub function of the Main function")
        
    return sub_function()
```

```python
@main_function
def Another_one(name,age):
    def inner():
        return f"Name: {name} Age: {age}"
    return inner
```

```python
def decorator_fun(func):
    def wrapper():
        print("Something will happen before the function")
        func()
        print("After the function")
    return wrapper # this will return the object only to the main function 

@decorator_fun
def greeting():
    print("Hello Noman How Are You?")
greeting()  
```

```python
# let say return wrapper return output 
# mean return wrapper() will apply for the same function
def decorator_fun(func):
    def wrapper():
        print("Something will happen before the function")
        func()
        print("After the function")
    return wrapper() # this enforce to return the output of this finction

@decorator_fun
def greeting():
    print("Hello Noman How Are You?")
# no need to call the greetin function
```

```python
def repeat(n):
    def decorator(fun):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                fun(*args,**kwargs)
        return wrapper()
    return decorator

@repeat(5)
def greeting():
    print("Hy Noman")
```
