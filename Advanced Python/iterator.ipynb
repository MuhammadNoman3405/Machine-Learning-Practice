
### Iterators

Iterators are advanced Python concepts that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
# let take a simple for loop iterator
my_list=[1,2,3,4,5,6]
for items in my_list:
    print(items)
```


**Generators**

Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
# locally we see this concept 
def square(n):
    for i in range(n + 1):
        print(f"{i**2}")
    
square(3)
```

```python
# return only gives one value return
def square(n):
    for i in range(n + 1):
        return i**2
    
square(3)
```

```python
# if we use the keyword yield then it will become the generator
def square_value(n):
    for i in range(n + 1):
        yield i**2
    
square_value(3)
```

```python
# how to access all elements one by one  stored in the generator
for items in square_value(3):
    print(items)
```

```python
# i can also access it viva next method
def square_value2(n):
    for i in range(n + 1):
        yield i**2
    
square_value2(3)
```

```python
a=square_value2(3)
print(next(a))
print(next(a))
print(next(a))
print(next(a))
```

```python
# i can i also get viva applying  the for loop
# i can also access it viva next method
def square_value3(n):
    for i in range(n + 1):
        yield i**2
    
square_value3(3)
```

```python
var=square_value3(3)
for i in var:
    print(i)
```

```python
def generate():
    yield 1 # yield similarly works as return
    yield 2
    yield 3
    
var=generate()
for i in var:
    print(i)
```

```python
# now it will throw an eror if i again want to print out its element
next(var)
```

```python
def generate1():
    yield 1 # yield similarly works as return
    yield 2
    yield 3
    
var=generate1()
var
```

```python
# try to practical example
def read_file(path):
    with open(path,'r') as file:
        for line in file:
            yield line
```

```python
file_path='large_file.txt'
for line in read_file(file_path):
    print(line.strip())
```
